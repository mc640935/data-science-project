```python
import numpy as np
import pyglet
import geoplotlib

from geoplotlib.core import BatchPainter
from geoplotlib.layers import BaseLayer
from geoplotlib.utils import (
    BoundingBox,
    DataAccessObject
)


europe_data_object = DataAccessObject(
    europe_data
)


europe_bbox = BoundingBox(
    north=70,
    west=-12,
    south=34,
    east=42
)


class FilterLayer(BaseLayer):

    def __init__(
        self,
        data,
        bbox
    ):
        self.data = data
        self.view = bbox

        self.start_letters = list(
            "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
        )

        self.start_letter = 0

        self.painter = BatchPainter()

    def invalidate(self, proj):

        current_letter = self.start_letters[
            self.start_letter
        ]

        mask = np.array([
            str(city).upper().startswith(
                current_letter
            )
            for city in self.data["City"]
        ])

        filtered_data = self.data.where(
            mask
        )

        self.painter = BatchPainter()

        x, y = proj.lonlat_to_screen(
            filtered_data["lon"],
            filtered_data["lat"]
        )

        self.painter.set_color(
            [255, 0, 0]
        )

        self.painter.points(
            x,
            y,
            4,
            rounded=True
        )

    def draw(
        self,
        proj,
        mouse_x,
        mouse_y,
        ui_manager
    ):
        self.painter.batch_draw()

        current_letter = self.start_letters[
            self.start_letter
        ]

        ui_manager.info(
            "Cities starting with: "
            + current_letter
        )

    def on_key_release(
        self,
        key,
        modifiers
    ):
        if key == pyglet.window.key.RIGHT:

            self.start_letter = (
                self.start_letter + 1
            ) % len(self.start_letters)

            return True

        if key == pyglet.window.key.LEFT:

            self.start_letter = (
                self.start_letter - 1
            ) % len(self.start_letters)

            return True

        return False

    def bbox(self):
        return self.view


geoplotlib.clear()

geoplotlib.tiles_provider(
    esri_tiles
)

geoplotlib.set_bbox(
    europe_bbox
)

geoplotlib.set_window_size(
    1100,
    750
)

geoplotlib.add_layer(
    FilterLayer(
        europe_data_object,
        europe_bbox
    )
)

geoplotlib.show()
```
